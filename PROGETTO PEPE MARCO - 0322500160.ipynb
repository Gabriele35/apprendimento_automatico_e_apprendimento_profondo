```python
#FASE 1 : Inizializzazione , importazione e studio dell'ambiente

import numpy as np
import pandas as pd
import sklearn
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()

print(data.data.shape)
print(data.target.shape)
print(data.feature_names)
print(data.target_names)

df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target

df.head()

print(df["target"].value_counts())

#FASE 2 : Parametrizzazione

X = df.drop("target", axis=1)
y = df["target"]

print("Dimensione X:", X.shape)
print("Dimensione y:", y.shape)

#FASE 3 : Suddivisione Training , Validation , Test

from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X, y,
    test_size=0.15,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.1765,
    random_state=42,
    stratify=y_temp
)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

#FASE 4 : Standardizzazione

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

print("Training:", X_train_scaled.shape)
print("Validation:", X_val_scaled.shape)
print("Test:", X_test_scaled.shape)

#FASE 5 : PCA

from sklearn.decomposition import PCA

pca = PCA()

X_train_pca = pca.fit_transform(X_train_scaled)

print("Numero di componenti:", pca.n_components_)
print("Varianza spiegata:")
print(pca.explained_variance_ratio_)

#FASE 6 : Definizione varianza

cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

for i, variance in enumerate(cumulative_variance, start=1):
    print(f"Componenti: {i:2d} - Varianza cumulativa: {variance:.2%}")


# Trasformiamo anche validation e test usando la PCA già calcolata
X_val_pca = pca.transform(X_val_scaled)
X_test_pca = pca.transform(X_test_scaled)

# FASE 7 : Grafico delle prime due componenti principali
plt.figure(figsize=(8, 6))

for classe, nome in zip([0, 1], data.target_names):
    plt.scatter(
        X_train_pca[y_train == classe, 0],
        X_train_pca[y_train == classe, 1],
        label=nome,
        alpha=0.7
    )

plt.xlabel("Prima componente principale (PC1)")
plt.ylabel("Seconda componente principale (PC2)")
plt.title("PCA - Prime due componenti principali")
plt.legend()
plt.grid(True)
plt.show()

# FASE 8 : Contributo delle caratteristiche alle prime due componenti

componenti = pd.DataFrame(
    pca.components_[:2],
    columns=X.columns,
    index=["PC1", "PC2"]
)

print("Contributo delle caratteristiche alla PC1:")
print(componenti.loc["PC1"].sort_values(key=abs, ascending=False).head(10))

print("\nContributo delle caratteristiche alla PC2:")
print(componenti.loc["PC2"].sort_values(key=abs, ascending=False).head(10))

#FASE 9 : Varianza cumulativa spiegata

plt.figure(figsize=(8, 5))

componenti_num = range(1, len(cumulative_variance) + 1)

plt.plot(componenti_num, cumulative_variance * 100, marker="o")

plt.xlabel("Numero di componenti principali")
plt.ylabel("Varianza spiegata cumulativa (%)")
plt.title("Varianza spiegata cumulativa dalla PCA")
plt.grid(True)

plt.axhline(y=95, linestyle="--", label="95%")
plt.legend()

plt.show()

#FASE 10 : Scelta algoritmi
#Gli algoritmi utilizzati saranno KNN , SVM e Random Forest , li scriviamo e li valutiamo

#ALGORITMO 1 : KNN

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)
print("Modello KNN addestrato.")

#Validation set KNN

from sklearn.metrics import accuracy_score

y_val_pred_knn = knn.predict(X_val_scaled)
accuracy_knn = accuracy_score(y_val, y_val_pred_knn)
print("Accuracy KNN sul validation set:", accuracy_knn)

#ALGORITMO 2 : SVM

from sklearn.svm import SVC

svm = SVC(kernel="rbf", probability=True, random_state=42)
svm.fit(X_train_scaled, y_train)
print("Modello SVM addestrato.")

#Validation set SVM

from sklearn.metrics import accuracy_score

y_val_pred_svm = svm.predict(X_val_scaled)
accuracy_svm = accuracy_score(y_val, y_val_pred_svm)
print("Accuracy SVM sul validation set:", accuracy_svm)

#ALGORTIMO 3 : Random Forest

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100,random_state=42)
rf.fit(X_train_scaled, y_train)
print("Modello Random Forest addestrato.")

#Validation set Random Forest

y_val_pred_rf = rf.predict(X_val_scaled)
accuracy_rf = accuracy_score(y_val, y_val_pred_rf)
print("Accuracy Random Forest sul validation set:", accuracy_rf)


#Adesso passiamo alle varie fasi ma prima classifichiamo benigni e maligni , ci permettera' meglio di trovare i cattivi
print("Classe 0:", data.target_names[0])
print("Classe 1:", data.target_names[1])

# FASE 11 : Adesso passiamo alla precisione di tutte valutazioni

from sklearn.metrics import (precision_score, recall_score, f1_score, roc_auc_score)

# Predizioni dei tre modelli
predizioni = {
    "KNN": y_val_pred_knn,
    "SVM": y_val_pred_svm,
    "Random Forest": y_val_pred_rf
}

# Calcolo delle metriche
for nome, y_pred in predizioni.items():
    accuracy = accuracy_score(y_val, y_pred)
    precision = precision_score(y_val, y_pred, pos_label=0)
    recall = recall_score(y_val, y_pred, pos_label=0)
    f1 = f1_score(y_val, y_pred, pos_label=0)
    
    print(f"\n--- {nome} ---")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-score : {f1:.4f}")

#FASE 12 : Creazione Confusion Matrix

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

#Algoritmo 1 KNN

cm_knn = confusion_matrix(y_val, y_val_pred_knn, labels=[0, 1])
disp = ConfusionMatrixDisplay(confusion_matrix=cm_knn,display_labels=["Malignant", "Benign"])

disp.plot()
plt.title("Confusion Matrix - KNN")
plt.show()
print("Confusion Matrix KNN:")
print(cm_knn)

#Algoritmo 2 : SVM

cm_svm = confusion_matrix(y_val, y_val_pred_svm, labels=[0, 1])
disp = ConfusionMatrixDisplay(confusion_matrix=cm_svm,display_labels=["Malignant", "Benign"])

disp.plot()
plt.title("Confusion Matrix - SVM")
plt.show()
print("Confusion Matrix SVM:")
print(cm_svm)

#Algoritmo 3 : RF

cm_rf = confusion_matrix(y_val, y_val_pred_rf, labels=[0, 1])

disp = ConfusionMatrixDisplay(confusion_matrix=cm_rf,display_labels=["Malignant", "Benign"])

disp.plot()
plt.title("Confusion Matrix - Random Forest")
plt.show()
print("Confusion Matrix Random Forest:")
print(cm_rf)

#FASE 13 : Controllo ROC AUC di tutti e 3 gli algoritmi

y_val_prob_knn = knn.predict_proba(X_val_scaled)[:, 1]
y_val_prob_svm = svm.predict_proba(X_val_scaled)[:, 1]
y_val_prob_rf = rf.predict_proba(X_val_scaled)[:, 1]

auc_knn = roc_auc_score(y_val, y_val_prob_knn)
auc_svm = roc_auc_score(y_val, y_val_prob_svm)
auc_rf = roc_auc_score(y_val, y_val_prob_rf)

print(f"ROC AUC KNN: {auc_knn:.4f}")
print(f"ROC AUC SVM: {auc_svm:.4f}")
print(f"ROC AUC Random Forest: {auc_rf:.4f}")

#FASE 14 : Calcolo curva ROC AUC sul Validation

from sklearn.metrics import roc_curve

fpr_knn, tpr_knn, _ = roc_curve(y_val, y_val_prob_knn)
fpr_svm, tpr_svm, _ = roc_curve(y_val, y_val_prob_svm)
fpr_rf, tpr_rf, _ = roc_curve(y_val, y_val_prob_rf)

plt.figure(figsize=(8, 6))

plt.plot(fpr_knn, tpr_knn, label=f"KNN (AUC = {auc_knn:.4f})")
plt.plot(fpr_svm, tpr_svm, label=f"SVM (AUC = {auc_svm:.4f})")
plt.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {auc_rf:.4f})")

plt.plot([0, 1], [0, 1], linestyle="--", label="Classificatore casuale")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Curve ROC - Confronto tra i classificatori")
plt.legend()
plt.grid(True)
plt.show()

#Abbiamo fatto i test sul Validation , adesso facciamo quelli sul Test
# FASE 15 Calcolo Valori sul Test , ROC e curva ROC

y_test_pred_knn = knn.predict(X_test_scaled)
y_test_pred_svm = svm.predict(X_test_scaled)
y_test_pred_rf = rf.predict(X_test_scaled)

print("Predizioni sul test set completate.")

for nome, y_pred in {
    "KNN": y_test_pred_knn,
    "SVM": y_test_pred_svm,
    "Random Forest": y_test_pred_rf
}.items():

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, pos_label=0)
    recall = recall_score(y_test, y_pred, pos_label=0)
    f1 = f1_score(y_test, y_pred, pos_label=0)

    print(f"\n--- {nome} ---")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-score : {f1:.4f}")

    
y_test_prob_knn = knn.predict_proba(X_test_scaled)[:, 1]
y_test_prob_svm = svm.predict_proba(X_test_scaled)[:, 1]
y_test_prob_rf = rf.predict_proba(X_test_scaled)[:, 1]

auc_test_knn = roc_auc_score(y_test, y_test_prob_knn)
auc_test_svm = roc_auc_score(y_test, y_test_prob_svm)
auc_test_rf = roc_auc_score(y_test, y_test_prob_rf)

print(f"ROC AUC KNN: {auc_test_knn:.4f}")
print(f"ROC AUC SVM: {auc_test_svm:.4f}")
print(f"ROC AUC Random Forest: {auc_test_rf:.4f}")

# Calcolo delle curve ROC sul test set
fpr_test_knn, tpr_test_knn, _ = roc_curve(y_test, y_test_prob_knn)
fpr_test_svm, tpr_test_svm, _ = roc_curve(y_test, y_test_prob_svm)
fpr_test_rf, tpr_test_rf, _ = roc_curve(y_test, y_test_prob_rf)


plt.figure(figsize=(8, 6))

plt.plot(
    fpr_test_knn,
    tpr_test_knn,
    label=f"KNN (AUC = {auc_test_knn:.4f})"
)

plt.plot(
    fpr_test_svm,
    tpr_test_svm,
    label=f"SVM (AUC = {auc_test_svm:.4f})"
)

plt.plot(
    fpr_test_rf,
    tpr_test_rf,
    label=f"Random Forest (AUC = {auc_test_rf:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Classificatore casuale"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Curve ROC - Test Set")
plt.legend()
plt.grid(True)
plt.show()

#FASE 16 Deep Learning , faremo tutte le operazioni che abbiamo effettuato precedentemente dai grafici , matrici e 

print("TensorFlow versione:", tf.__version__)

from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.Input(shape=(X_train_scaled.shape[1],)),
    layers.Dense(16, activation="relu"),
    layers.Dense(8, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])

model.summary()

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Modello compilato correttamente.")

#FASE 17 Addestramento DL

history = model.fit(
    X_train_scaled,
    y_train,
    validation_data=(X_val_scaled, y_val),
    epochs=50,
    batch_size=16,
    verbose=1
)

#Grafico validation
plt.figure(figsize=(8, 5))

plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")

plt.xlabel("Epoca")
plt.ylabel("Accuracy")
plt.title("Accuracy durante l'addestramento della rete neurale")
plt.legend()
plt.grid(True)
plt.show()

#Grafico loss
plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training")
plt.plot(history.history["val_loss"], label="Validation")

plt.xlabel("Epoca")
plt.ylabel("Loss")
plt.title("Loss durante l'addestramento della rete neurale")
plt.legend()
plt.grid(True)
plt.show()

test_loss, test_accuracy = model.evaluate(
    X_test_scaled,
    y_test,
    verbose=0
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)

y_test_prob_dl = model.predict(X_test_scaled, verbose=0)
y_test_pred_dl = (y_test_prob_dl >= 0.5).astype(int).ravel()

cm_dl = confusion_matrix(y_test, y_test_pred_dl)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_dl,
    display_labels=data.target_names
)

disp.plot()
plt.title("Matrice di confusione - Rete neurale")
plt.show()

accuracy_dl = accuracy_score(y_test, y_test_pred_dl)
precision_dl = precision_score(y_test, y_test_pred_dl, pos_label=0)
recall_dl = recall_score(y_test, y_test_pred_dl, pos_label=0)
f1_dl = f1_score(y_test, y_test_pred_dl, pos_label=0)

print("Accuracy:", accuracy_dl)
print("Precision:", precision_dl)
print("Recall:", recall_dl)
print("F1-score:", f1_dl)

auc_dl = roc_auc_score(y_test, y_test_prob_dl)

fpr_dl, tpr_dl, _ = roc_curve(y_test, y_test_prob_dl)

print("ROC AUC:", auc_dl)

plt.figure(figsize=(8, 6))
plt.plot(fpr_dl, tpr_dl, label=f"Rete neurale (AUC = {auc_dl:.4f})")
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Curva ROC - Rete neurale")
plt.legend()
plt.grid(True)
plt.show()

#FASE 18 : Grafico tra tutti gli algoritmi e la rete neurale

modelli = ["KNN", "SVM", "Random Forest", "Rete neurale"]

accuracy = [0.9651, 0.9651, 0.9302, 0.9419]
f1 = [0.9524, 0.9538, 0.9091, 0.9254]
roc_auc = [0.9887, 0.9936, 0.9931, 0.9936]

x = np.arange(len(modelli))
larghezza = 0.25

plt.figure(figsize=(10, 6))

plt.bar(x - larghezza, accuracy, larghezza, label="Accuracy")
plt.bar(x, f1, larghezza, label="F1-score")
plt.bar(x + larghezza, roc_auc, larghezza, label="ROC AUC")

plt.xticks(x, modelli)
plt.ylabel("Valore")
plt.title("Confronto delle prestazioni dei modelli")
plt.ylim(0.85, 1.02)
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```
